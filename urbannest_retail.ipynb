```python
urbannest_df_copy = urbannest_df.copy()
```

```python
# Trim whitespace in text columns.
text_cols = urbannest_df_copy.select_dtypes(include="str").columns
urbannest_df_copy[text_cols] = urbannest_df_copy[text_cols].apply(
    lambda column: column.str.strip()
)
```

```python
# Standardize known gender values; treat "Unknown" as missing.
gender = urbannest_df_copy["gender"].str.casefold()
urbannest_df_copy["gender"] = gender.map({
    "m": "Male",
    "male": "Male",
    "f": "Female",
    "female": "Female",
    "unknown": pd.NA,
})
```

```python
# Normalize city spelling and capitalization.
city = urbannest_df_copy["city"].str.casefold()
urbannest_df_copy["city"] = (
    city.replace({"portharcourt": "port harcourt"}).str.title()
)
```

```python
urbannest_df_copy.head()
```

```python
# Standardize inconsistent category values.
status = urbannest_df_copy["order_status"].str.casefold()
urbannest_df_copy["order_status"] = (
    status.replace({"cancelled": "Cancelled"}).str.title()
)

payment = urbannest_df_copy["payment_method"].str.casefold()
urbannest_df_copy["payment_method"] = (
    payment.replace({"CARD": "Card"}).str.title()
)

sales = urbannest_df_copy["sales_channel"].str.casefold()
urbannest_df_copy["sales_channel"] = (
    sales.replace({"mobile app": "Mobile App"}).str.title()
)

status = urbannest_df_copy["delivery_status"].str.casefold()
urbannest_df_copy["delivery_status"] = (
    status.replace({"delivered": "Delivered"}).str.title()
)

flag = urbannest_df_copy["return_flag"].str.casefold()
urbannest_df_copy["return_flag"] = (
    flag.replace({"NO": "No", "YES": "Yes"}).str.title()
)
```

```python
# Add subtotal column for analysis.
urbannest_df_copy["subtotal"] = urbannest_df_copy["quantity"] * urbannest_df_copy["unit_price"]
```

```python
#
urbannest_df_copy.insert(
    urbannest_df_copy.columns.get_loc("unit_price") + 1,
    "subtotal",
    urbannest_df_copy.pop("subtotal")
)
```

```python
# Display prices with thousands separators without changing the original numeric column.
urbannest_df_copy.assign(
    unit_price=lambda df: df["unit_price"].map("{:,.0f}".format),
    order_revenue=lambda df: df["order_revenue"].map("{:,.0f}".format),
    delivery_fee=lambda df: df["delivery_fee"].map("{:,.0f}".format),
    subtotal=lambda df: df["subtotal"].map("{:,.0f}".format)
)
```

```python
# Convert data type of date column to datetime.
urbannest_df_copy["order_date"] = pd.to_datetime(
    urbannest_df_copy["order_date"], errors="coerce"
)
```

```python
# Add order_month column for analysis, and use month names.
urbannest_df_copy["order_month"] = urbannest_df_copy["order_date"].dt.month_name()
```

```python
# Add order_year columns for analysis.
urbannest_df_copy["order_year"] = urbannest_df_copy["order_date"].dt.year
```

```python

# Rearrange columns to place order_month and order_year after order_date.
columns = [
    column
    for column in urbannest_df_copy.columns
    if column not in ("order_month", "order_year")
]
date_index = columns.index("order_date")
columns[date_index + 1 : date_index + 1] = ["order_month", "order_year"]
urbannest_df_copy = urbannest_df_copy[columns]
```

```python
# Convert data types for date and phone number columns.
urbannest_df_copy["phone_number"] = urbannest_df_copy["phone_number"].astype("string")
```

```python
# Mask phone numbers to show only the first three and last two digits, replacing the middle digits with asterisks.
urbannest_df_copy["phone_number"] = urbannest_df_copy["phone_number"].str.replace(r"^(\d{3})\d+(\d{2})$", lambda m: m.group(1) + "*" * (len(m.group(0)) - 5) + m.group(2), regex=True)
```
